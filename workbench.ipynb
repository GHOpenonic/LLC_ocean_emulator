```python

theta =  llc.Theta.isel(time=0,k=0,face=1)


theta_coarse_third = theta.coarsen(j=16, i=16, boundary="trim").mean()  # NaN-aware mean; all-land boxes stay NaN
theta_coarse_third = theta_coarse_third.compute()

theta_coarse_one = theta.coarsen(j=48, i=48, boundary="trim").mean()  # NaN-aware mean; all-land boxes stay NaN
theta_coarse_one = theta_coarse_one.compute()
```

```python
fig, axes = plt.subplots(3,4,figsize=(14, 16), dpi=100, layout='tight')
colormap='bwr'
ax=axes[0,0]
cf = ax.imshow(theta,cmap=colormap)
ax.set_title(f'surface theta sharp (1/48)')

ax=axes[1,0]
cf = ax.imshow(theta_coarse_third,cmap=colormap)
ax.set_title(f'surface theta coarse (1/3)')

ax=axes[2,0]
cf=ax.imshow(theta_coarse_one,cmap=colormap)
ax.set_title(f'surface theta coarse (1 deg)')

ax=axes[0,1]
cf = ax.imshow(theta.isel(i=slice(2880,3600),j=slice(720,1440)),cmap=colormap)
ax.set_title(f'surface theta sharp (1/48) zoom')
ax=axes[1,1]
cf = ax.imshow(theta_coarse_third.isel(i=slice(2880//16,3600//16),j=slice(720//16,1440//16)),cmap=colormap)
ax.set_title(f'surface theta coarse (1/3) zoom')
ax=axes[2,1]
cf=ax.imshow(theta_coarse_one.isel(i=slice(2880//48,3600//48),j=slice(720//48,1440//48)),cmap=colormap)
ax.set_title(f'surface theta coarse (1 deg) zoom')

ax=axes[0,2]
cf = ax.imshow(theta.isel(i=slice(2880+100,2880+300),j=slice(720+100,720+300)),cmap=colormap)
ax.set_title(f'surface theta sharp (1/48) zoom2')
ax=axes[1,2]
cf = ax.imshow(theta_coarse_third.isel(i=slice((2880+100)//16,(2880+300)//16),j=slice((720+100)//16,(720+300)//16)),cmap=colormap)
ax.set_title(f'surface theta coarse (1/3) zoom2')
ax=axes[2,2]
cf=ax.imshow(theta_coarse_one.isel(i=slice((2880+100)//48,(2880+300)//48),j=slice((720+100)//48,(720+300)//48)),cmap=colormap)
ax.set_title(f'surface theta coarse (1 deg) zoom2')

ax=axes[0,3]
cf = ax.imshow(theta.isel(i=slice(2880+100+25,2880+100+25+50),j=slice(720+100+75,720+100+75+50)),cmap=colormap)
ax.set_title(f'surface theta sharp (1/48) zoom3')

ax=axes[1,3]
cf = ax.imshow(theta_coarse_third.isel(i=slice((2880+100+25)//16,(2880+100+25+48)//16),j=slice((720+100+75)//16,(720+100+75+48)//16)),cmap=colormap)
ax.set_title(f'surface theta coarse (1/3) zoom3')

ax=axes[2,3]
cf=ax.imshow(theta_coarse_one.isel(i=slice((2880+100+25)//48,(2880+100+25+48)//48),j=slice((720+100+75)//48,(720+100+75+48)//48)),cmap=colormap)
ax.set_title(f'surface theta coarse (1 deg) zoom3')
```
